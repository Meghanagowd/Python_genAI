Shuffle manually (no built-in)

```python
import random

my_list = [1,2,3,4,5]

for x in range(0,len(my_list)):
    i = random.randint(0,len(my_list)-1)
    j = random.randint(0,len(my_list)-1)
    my_list[i],my_list[j] = my_list[j] , my_list[i]

print(my_list)

```

Password Generator

```python
import random

def generate_password(l,n,s):
    letters = 'abcdefghijklmnopqrstuvwxyz'
    numbers = '0123456789'
    symbols = '!@#$%^&*'
    result = []

    for i in range(l):
        x = random.randint(0,len(letters)-1)
        result.append(letters[x])
    for j in range(n):
        x = random.randint(0,len(numbers)-1)
        result.append(numbers[x])    
    for i in range(s):
        x = random.randint(0,len(symbols)-1)
        result.append(symbols[x])    

    random.shuffle(result)
    return "".join(result)


print(generate_password(3,2,2))


```

Password strength checker

```python
def check_password(p):
    has_num = any(ch.isdigit() for ch in p)
    has_symbol = any(ch in '!@#$%^&*' for ch in p)

    if len(p) > 6 and has_num and has_symbol:
        return 'strong'
    else :
        return "weak"
    
p = "abc234!@#$"    
print(check_password(p))    
```
